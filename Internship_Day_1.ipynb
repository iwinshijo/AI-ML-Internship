```python
df['Age_mean_imputed'] = df['Age'].fillna(df['Age'].mean())
df['Age_median_imputed'] = df['Age'].fillna(df['Age'].median())
df['Cabin_mode_imputed'] = df['Cabin'].fillna(df['Cabin'].mode()[0])
df[['Age','Age_mean_imputed','Age_median_imputed','Cabin','Cabin_mode_imputed']]
```

```python
df
```

```python
sns.set(style='whitegrid')

plt.figure(figsize=(10,6))
sns.boxplot(data=df,x='Pclass',y='Age',hue='Pclass',palette='pastel', legend=False)
plt.title('Boxplot of Age by Passenger class',fontsize=16)
plt.xlabel('Passenger class',fontsize=12)
plt.ylabel('Age',fontsize=12)
plt.show()
```

```python
sns.set(style='whitegrid')

plt.figure(figsize=(10,6))
sns.boxplot(data=df,x='Pclass',y='Survived',hue='Pclass',palette='pastel', legend=False)
plt.title('Boxplot of Survived by Passenger class',fontsize=16)
plt.xlabel('Passenger class',fontsize=12)
plt.ylabel('Survived',fontsize=12)
plt.show()
```

```python
sns.set(style='whitegrid')

plt.figure(figsize=(10,6))
sns.boxplot(data=df,x='Embarked',y='Fare',palette='pastel', legend=False)
plt.title('Boxplot of Age by Embarek class',fontsize=16)
plt.xlabel('Embarked class',fontsize=12)
plt.ylabel('Fare',fontsize=12)
plt.show()
```

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

encoded_data=df.copy()
label_encoder=LabelEncoder()
encoded_data['Sex']=label_encoder.fit_transform(encoded_data['Sex'])

one_hot_encoded=pd.get_dummies(encoded_data['Embarked'],prefix='Embarked')
encoded_data=pd.concat([encoded_data,one_hot_encoded],axis=1)

pclass_mapping={1:'First',2:'Second',3:'Third'}
encoded_data['Pclass_Category']=encoded_data['Pclass'].map(pclass_mapping)

print(encoded_data[['Sex','Embarked','Embarked_Q', 'Embarked_S','Embarked_C','Pclass']])
```
